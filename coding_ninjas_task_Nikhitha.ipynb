## Part B — Data Preprocessing

The dataset was checked for incorrect data types, missing values, duplicate records,
and suspicious values. Necessary corrections were made to prepare the dataset
for analysis and machine learning.

```python
# Convert horsepower from text/object to numeric
df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")

# Check the updated data type
print("Horsepower data type:", df["horsepower"].dtype)

# Check whether conversion created any missing values
print("Missing horsepower values:", df["horsepower"].isnull().sum())
```

```python
# Check all data types after preprocessing
df.info()
```

```python
# Fill missing horsepower values using the median
horsepower_median = df["horsepower"].median()

df["horsepower"] = df["horsepower"].fillna(horsepower_median)

print("Median horsepower used:", horsepower_median)
print("Missing horsepower values after filling:",
      df["horsepower"].isnull().sum())
```

### Handling Missing Horsepower Values

After converting the `horsepower` column to numeric format, 6 missing values were detected.

These missing values were replaced with the median horsepower value. The median was chosen because it is less affected by unusually high or low horsepower values than the mean.

```python
# Final preprocessing check
print("Final dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

# Task 2 — Linear Regression

The goal of this task is to build a Linear Regression model that predicts MPG
using vehicle characteristics from the cleaned Auto MPG dataset.

## Part A — Data Preparation

```python
# Load the cleaned dataset
clean_df = pd.read_csv("cleaned_auto_mpg.csv")

clean_df.head()
```

```python
print("Dataset shape:", clean_df.shape)
print("\nData types:")
print(clean_df.dtypes)
```

```python
# Target variable
y = clean_df["mpg"]

# Features
X = clean_df.drop(columns=["mpg", "car_name"])

print("Features used:")
print(X.columns)
```

```python
# Convert origin into one-hot encoded numerical features
X = pd.get_dummies(X, columns=["origin"], drop_first=True)

X.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

## Part B — Linear Regression

### Model 1: Predicting MPG Using Vehicle Weight

A simple Linear Regression model is first built using only vehicle weight
to understand how strongly weight alone can predict MPG.

```python
from sklearn.linear_model import LinearRegression

# Use only weight as the feature
X_weight = clean_df[["weight"]]
y_weight = clean_df["mpg"]

Xw_train, Xw_test, yw_train, yw_test = train_test_split(
    X_weight,
    y_weight,
    test_size=0.20,
    random_state=42
)

weight_model = LinearRegression()

weight_model.fit(Xw_train, yw_train)

weight_predictions = weight_model.predict(Xw_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

weight_mae = mean_absolute_error(yw_test, weight_predictions)
weight_mse = mean_squared_error(yw_test, weight_predictions)
weight_rmse = np.sqrt(weight_mse)
weight_r2 = r2_score(yw_test, weight_predictions)

print("Weight-only Model Performance")
print("MAE:", weight_mae)
print("MSE:", weight_mse)
print("RMSE:", weight_rmse)
print("R²:", weight_r2)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(Xw_test["weight"], yw_test, label="Actual MPG")
plt.scatter(Xw_test["weight"], weight_predictions, label="Predicted MPG")

plt.xlabel("Vehicle Weight")
plt.ylabel("MPG")
plt.title("Weight vs MPG — Linear Regression")
plt.legend()

plt.show()
```

### Interpretation of the Weight-Only Model

The weight-only Linear Regression model achieved an R² score of approximately 0.723,
which means that vehicle weight alone explains about 72.3% of the variation in MPG
in the test data.

The MAE is approximately 3.12 MPG, meaning that the model's predictions differ
from the actual MPG by about 3.12 MPG on average.

The RMSE is approximately 3.86 MPG. Since RMSE gives more importance to larger errors,
it indicates that some predictions have larger deviations from the actual values.

Overall, vehicle weight alone is a strong predictor of MPG, but additional vehicle
characteristics may improve the prediction performance.

### Model 2: Multiple Linear Regression

A multiple Linear Regression model is built using several relevant vehicle
characteristics to determine whether combining multiple features improves MPG prediction.

```python
# Build the multiple-feature Linear Regression model
multi_model = LinearRegression()

multi_model.fit(X_train, y_train)

multi_predictions = multi_model.predict(X_test)
```

```python
multi_mae = mean_absolute_error(y_test, multi_predictions)
multi_mse = mean_squared_error(y_test, multi_predictions)
multi_rmse = np.sqrt(multi_mse)
multi_r2 = r2_score(y_test, multi_predictions)

print("Multiple Linear Regression Performance")
print("MAE:", multi_mae)
print("MSE:", multi_mse)
print("RMSE:", multi_rmse)
print("R²:", multi_r2)
```

```python
print("Weight-only R²:", weight_r2)
print("Multiple-feature R²:", multi_r2)

print("\nWeight-only RMSE:", weight_rmse)
print("Multiple-feature RMSE:", multi_rmse)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(y_test, multi_predictions)

plt.xlabel("Actual MPG")
plt.ylabel("Predicted MPG")
plt.title("Actual vs Predicted MPG")

# Ideal prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.show()
```

### Comparison of the Two Models

The multiple-feature Linear Regression model performed better than the weight-only model.

The R² score increased from approximately 0.723 to 0.845, indicating that the multiple-feature model explains a larger proportion of the variation in MPG.

The RMSE decreased from approximately 3.86 MPG to 2.89 MPG, showing that the multiple-feature model makes smaller prediction errors overall.

This suggests that MPG is influenced by multiple vehicle characteristics rather than vehicle weight alone.

```python
# Compare training and testing performance
train_r2 = multi_model.score(X_train, y_train)
test_r2 = multi_model.score(X_test, y_test)

print("Training R²:", train_r2)
print("Testing R²:", test_r2)
```

```python
train_predictions = multi_model.predict(X_train)

train_rmse = np.sqrt(mean_squared_error(y_train, train_predictions))
test_rmse = np.sqrt(mean_squared_error(y_test, multi_predictions))

print("Training RMSE:", train_rmse)
print("Testing RMSE:", test_rmse)
```

## Part C — Model Analysis

### Training vs Testing Performance

The model achieved an R² score of approximately 0.819 on the training data and
0.845 on the testing data.

The training RMSE was approximately 3.37 MPG, while the testing RMSE was
approximately 2.89 MPG.

The training and testing performances are relatively close, which suggests that
the model generalizes well to unseen data.

There is no strong evidence of overfitting, because the model does not perform
much better on the training data than on the testing data.

There is also no clear evidence of severe underfitting, since the model explains
a substantial portion of the variation in MPG.

### Interpretation of Evaluation Metrics

- **MAE (Mean Absolute Error):** Represents the average absolute difference between
  the actual MPG and predicted MPG.

- **MSE (Mean Squared Error):** Measures the average squared prediction error and
  gives more penalty to larger mistakes.

- **RMSE (Root Mean Squared Error):** Represents prediction error in the same unit
  as MPG, making it easier to interpret.

- **R² Score:** Indicates how much of the variation in MPG is explained by the model.
  The multiple-feature model achieved an R² of approximately 0.845, meaning it
  explains about 84.5% of the variation in MPG in the test set.
